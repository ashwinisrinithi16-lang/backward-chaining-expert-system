Rule-Based Expert System using Backward Chaining

```python
facts = {
    "GoodAttendance",
    "GoodMarks",
    "CompletedAssignments",
    "RegularPractice"
}

rules = {
    "EligibleForPlacement": [
        ["GoodAttendance", "GoodMarks", "CompletedAssignments"]
    ],

    "GoodAcademicPerformance": [
        ["GoodMarks", "RegularPractice"]
    ],

    "NeedsExtraPractice": [
        ["GoodAttendance", "CompletedAssignments"]
    ],

    "ReadyForInterview": [
        ["EligibleForPlacement", "RegularPractice"]
    ]
}

def backward_chaining(goal, facts, rules, steps):

    if goal in facts:
        steps.append("Fact found: " + goal)
        return True

    if goal not in rules:
        steps.append("No rule found for: " + goal)
        return False

    for conditions in rules[goal]:
        steps.append("To prove " + goal + ", checking: " + ", ".join(conditions))

        result = True

        for condition in conditions:
            if not backward_chaining(condition, facts, rules, steps):
                result = False
                break

        if result:
            steps.append("Rule satisfied: " + goal)
            return True

    steps.append("Cannot prove: " + goal)
    return False


goal = "EligibleForPlacement"

steps = []

result = backward_chaining(goal, facts, rules, steps)

print("BACKWARD CHAINING")
print("-----------------")
print("Goal:", goal)

print("\nReasoning Steps:")

for step in steps:
    print(step)

print("\nFinal Conclusion:")

if result:
    print(goal, "is proved.")
else:
    print(goal, "cannot be proved.")
```

Backward Chaining starts with a goal and works backward to find the facts needed to prove that goal.

For example:

Goal
  ↓
EligibleForPlacement
  ↓
GoodAttendance
GoodMarks
CompletedAssignments
  ↓
All facts are true
  ↓
Goal is proved

Facts: GoodAttendance, GoodMarks, CompletedAssignments, RegularPractice

Rules:

GoodAttendance + GoodMarks + CompletedAssignments → EligibleForPlacement
GoodMarks + RegularPractice → GoodAcademicPerformance
GoodAttendance + CompletedAssignments → NeedsExtraPractice
EligibleForPlacement + RegularPractice → ReadyForInterview
